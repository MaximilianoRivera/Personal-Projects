<h2 align = 'center'>Wine Classifier W/Naive Baiyes <h2>

```python
import pandas as pd
from sklearn.datasets import load_wine
```

```python
dir(load_wine())
```

```python
wine = pd.DataFrame(load_wine().data,columns=load_wine().feature_names)
wine
```

```python
wine_class = load_wine().target
wine_class
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test= train_test_split(wine,wine_class,test_size=0.25)
```

<h4>GaussianNB Classifier<h4>

```python
from sklearn.naive_bayes import GaussianNB
GNB_model = GaussianNB()
```

```python
GNB_model.fit(X_train,y_train)
```

```python
GNB_model.score(X_test,y_test)
```

```python
GNB_model.predict(X_test)
```

<h4> MultinomialNB Classifer <h4>

```python
from sklearn.naive_bayes import MultinomialNB
MNB_model = MultinomialNB()
MNB_model.fit(X_train,y_train)
```

```python
MNB_model.score(X_test,y_test)
```

```python
MNB_model.predict(X_test)
```

We can observe that the GaussianNB is the better option for the wine classification with a 100% accurracy
