```python
import pandas as pd
import numpy as np
```

```python
heart_disease = pd.read_csv('heart.csv')
heart_disease.head()
```

```python
heart_disease.shape
```

```python
heart_disease.describe()
```

```python
outlier_col = ['RestingBP','Cholesterol','MaxHR','Oldpeak']
```

```python
for rem_outlier in outlier_col:
    heart_disease[rem_outlier+' zscore']= (heart_disease[rem_outlier]-heart_disease[rem_outlier].mean())/heart_disease[rem_outlier].std()

heart_disease.head()
```

```python
heart_disease_no_outlier = heart_disease
```

```python
for rem_outlier in outlier_col:
    heart_disease_no_outlier = heart_disease_no_outlier[(heart_disease_no_outlier[rem_outlier+' zscore']>-3)&(heart_disease_no_outlier[rem_outlier+' zscore']<3)] 

heart_disease_no_outlier.describe()
```

```python
heart_disease_no_outlier.head()
```

```python
heart_disease_no_outlier['ST_Slope'].unique()
```

```python
dfd = pd.get_dummies(heart_disease_no_outlier,
                     columns=['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope'],
                     drop_first=True,
                     dtype=float
                     )
dfd.head()
```

```python
X = dfd.drop(['HeartDisease','RestingBP zscore','Cholesterol zscore','MaxHR zscore','Oldpeak zscore'],axis='columns')
X
```

```python
Y = dfd['HeartDisease']
Y.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
x_scaled = StandardScaler().fit_transform(X)
x_scaled
```

```python
from sklearn import svm
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x_scaled,Y,test_size=0.2)
```

```python
X_train.shape
```

```python
model = svm.SVC(gamma='auto',C=1,kernel='rbf')
model.fit(X_train,Y_train)
model.score(X_test,Y_test)
```

```python
c_score = cross_val_score(model,x_scaled,Y,cv=5)
c_score.mean()
```

```python
from sklearn.ensemble import BaggingClassifier
```

```python
clf = BaggingClassifier(
   estimator=model,
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=10
)
clf.fit(x_scaled,Y)
clf.oob_score_
```

```python
c_score_2 = cross_val_score(clf,x_scaled,Y,cv=5)
c_score_2.mean()
```

```python
from sklearn.tree import DecisionTreeClassifier
c_score_3 = cross_val_score(DecisionTreeClassifier(),x_scaled,Y,cv=5)
c_score_3.mean()
```

```python
clf_2 = BaggingClassifier(
    estimator= DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=10
)
```

```python
c_score_4 = cross_val_score(clf_2,x_scaled,Y,cv=5)
c_score_4.mean()
```
