```python
import pandas as pd
import numpy as np
```

```python
heart_disease = pd.read_csv('heart.csv')
heart_disease.head()
```

```python
heart_disease.shape
```

```python
heart_disease.describe()
```

```python
outlier_col = ['RestingBP','Cholesterol','MaxHR','Oldpeak']
```

```python
for rem_outlier in outlier_col:
    heart_disease[rem_outlier+' zscore']= (heart_disease[rem_outlier]-heart_disease[rem_outlier].mean())/heart_disease[rem_outlier].std()

heart_disease.head()
```

```python
heart_disease_no_outlier = heart_disease
```

```python
for rem_outlier in outlier_col:
    heart_disease_no_outlier = heart_disease_no_outlier[(heart_disease_no_outlier[rem_outlier+' zscore']>-3)&(heart_disease_no_outlier[rem_outlier+' zscore']<3)] 

heart_disease_no_outlier.describe()
```

```python
heart_disease_no_outlier.head()
```

```python
heart_disease_no_outlier['ST_Slope'].unique()
```

```python
dfd = pd.get_dummies(heart_disease_no_outlier,
                     columns=['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope'],
                     drop_first=True,
                     dtype=float
                     )
dfd.head()
```

```python
X = dfd.drop(['HeartDisease','RestingBP zscore','Cholesterol zscore','MaxHR zscore','Oldpeak zscore'],axis='columns')
X
```

```python
Y = dfd['HeartDisease']
Y
```

```python
from sklearn.preprocessing import StandardScaler
x_scaled = StandardScaler().fit_transform(X)
x_scaled
```

```python
from sklearn import svm
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
```

```python
model_params = {
    'svm': {
        'model': svm.SVC(gamma='auto'),
        'params': {
            'C':[1,5,10],
            'kernel': ['rbf','linear']
        }
    },

    'RandomForest':{
        'model': RandomForestClassifier(),
        'params':{
            'n_estimators':[1,5,10]
        }
    },

    'LogisticRegression': {
        'model': LogisticRegression(solver='liblinear',multi_class='auto'),
        'params':{
            'C':[1,5,10]
        }
    },
}
```

```python
scores = []

for model_name, mp in model_params.items():
    clf = GridSearchCV(mp['model'],mp['params'],cv=5,return_train_score=False)
    clf.fit(x_scaled,Y)
    scores.append({
        'model':model_name,
        'best_score':clf.best_score_,
        'best_params': clf.best_params_
    })
```

```python
df_scores = pd.DataFrame(scores,columns=['model','best_score','best_params'])
df_scores
```

```python
from sklearn.decomposition import PCA
pca = PCA(0.95)
x_pca = pca.fit_transform(x_scaled)
x_pca.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x_pca,Y,test_size=0.2)
```

```python
X_train.shape
```

```python
model = svm.SVC(gamma='auto',C=1,kernel='rbf')
model.fit(X_train,Y_train)
model.score(X_test,Y_test)
```

```python
clf = GridSearchCV(svm.SVC(gamma='auto'),{'C':[1,5,10],'kernel':['rbf']},cv=5,return_train_score=False)
clf.fit(x_pca,Y)
clf.best_score_
```
